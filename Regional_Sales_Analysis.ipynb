# 4. Data Cleaning

```python
# Dropping the Row ID Column, if we want to use Machine Learning
# df.drop('Row ID', axis=1, inplace=True)

# Converting the date column to datetime format (dd/mm/yyyy)
df['Order Date'] = pd.to_datetime(df['Order Date'], format='%d/%m/%Y')
df['Ship Date'] = pd.to_datetime(df['Ship Date'], format='%d/%m/%Y')
```

```python
# Converting the 'Postal Code' column to numeric format
df['Postal Code'] = df['Postal Code'].astype('Int64')
```

```python
# Understanding the rows with null value
df[df.isnull().any(axis=1)]
```

```python
# We can see that the postal code is not mentioned only for Burlington city in Vermont state. 
# So, we need to fill the postal code of that city.
df['Postal Code'] = df['Postal Code'].fillna(5401)
```

```python
df.isnull().sum()
```

```python
# Find out how often this “false duplicate” occurs
columns_to_check = [col for col in df.columns if col != 'Row ID']
duplicate_counts = df.duplicated(subset=columns_to_check, keep=False).sum()

print(f"Number of rows flagged as duplicates by the system: {duplicate_counts}")
```

```python
# If the number is very small (< 1–2% of the total data), 
# safeguard your analysis by removing it outright (keep=‘first’) 
# since it has no significant impact on total revenue but helps eliminate noise.
df = df.drop_duplicates(subset=columns_to_check, keep='first')
```

```python
df = df.reset_index(drop=True)
```

```python
# Converting the 'Postal Code' column to string format
df['Postal Code'] = df['Postal Code'].astype(str)
```

```python
df.info()    
```

```python
# Check the columns
df.columns
```

```python
# Changes the columns' name into snake_case type
df.columns = df.columns.str.lower()
df.columns = df.columns.str.replace(' ', '_')
df = df.rename(columns={'sub-category':'sub_category'})
```
